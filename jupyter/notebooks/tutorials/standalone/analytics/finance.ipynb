This tutorial consolidates revenue across legal entities that use different local rules, then removes intercompany activity from the group result.

The important lesson is ordering: calculate each leaf under its own policy first, add the leaves second, and apply eliminations last. Shared mappings and rule features keep the country-specific differences in data rather than separate formula copies.

This advanced tutorial assumes the relationship model from [E-commerce](https://featuremesh.com/docs/tutorials/analytics/ecomm) and persisted features from [SaaS metrics](https://featuremesh.com/docs/tutorials/analytics/saas). - Germany Unit includes shipping and excludes returns.

## Build the hierarchy and mappings

Declare the consolidation entities and the month input used by every query:


```python
%%featureql --client client

CREATE OR REPLACE FEATURES IN FM.FINANCE AS
PERSIST
    fc_entities := ENTITY(),
    entity_id := INPUT(BIGINT#fc_entities),
    transactions := ENTITY(),
    tx_id := INPUT(BIGINT#transactions),
    interco := ENTITY(),
    interco_id := INPUT(BIGINT#interco),
    month_input := INPUT(VARCHAR)
;
```

Then map entity metadata, local rules, transactions, and intercompany rows:


```python
%%featureql --client client

CREATE OR REPLACE FEATURES IN FM.FINANCE AS
PERSIST
    tables.dim_entities := EXTERNAL_COLUMNS(
        entity_id BIGINT#fc_entities BIND TO entity_id,
        name VARCHAR,
        parent_id BIGINT,
        currency VARCHAR
        FROM TABLE(tutorial_finance.entities)
    ),
    tables.rules := EXTERNAL_COLUMNS(
        entity_id BIGINT#fc_entities BIND TO entity_id,
        includes_shipping BOOLEAN,
        includes_returns BOOLEAN
        FROM TABLE(tutorial_finance.revenue_rules)
    ),
    tables.fct_tx := EXTERNAL_COLUMNS(
        tx_id BIGINT#transactions BIND TO tx_id,
        tx_entity_id BIGINT#fc_entities,
        account VARCHAR,
        amount DOUBLE,
        tx_date DATE
        FROM TABLE(tutorial_finance.transactions)
    ),
    tables.interco := EXTERNAL_COLUMNS(
        interco_id BIGINT#interco BIND TO interco_id,
        from_entity_id BIGINT#fc_entities,
        to_entity_id BIGINT#fc_entities,
        account VARCHAR,
        amount DOUBLE,
        tx_date DATE
        FROM TABLE(tutorial_finance.intercompany)
    )
;
```

Persist the reusable revenue building blocks, including the `MONTH_INPUT` used by each monthly query:


```python
%%featureql --client client

CREATE OR REPLACE FEATURES IN FM.FINANCE AS
PERSIST
    entity_name := tables.dim_entities[name],
    includes_shipping := tables.rules[includes_shipping],
    includes_returns := tables.rules[includes_returns],
    tx_account := tables.fct_tx[account],
    tx_amount := tables.fct_tx[amount],
    tx_month := DATE_FORMAT(tables.fct_tx[tx_date]::TIMESTAMP, '%Y-%m'),
    tx_entity_id := tables.fct_tx[tx_entity_id],
    ic_month := DATE_FORMAT(tables.interco[tx_date]::TIMESTAMP, '%Y-%m')
;
```

The library keeps local policy close to the entity it governs. The consolidation queries can therefore focus on rollup rather than repeat shipping and return conditions.

## Apply local rules in January

Bind January and calculate revenue independently for France and Germany:


```python
%%featureql --client client

DEFINE
    gross := COALESCE(
        entity_id.RELATED(
            SUM(tx_amount)
            FILTER (WHERE tx_account = 'gross_sales' AND tx_month = month_input)
            GROUP BY (tx_entity_id)
        ),
        0e0
    ),
    shipping := COALESCE(
        entity_id.RELATED(
            SUM(tx_amount)
            FILTER (WHERE tx_account = 'shipping_revenue'
            AND tx_month = month_input)
            GROUP BY (tx_entity_id)
        ),
        0e0
    ),
    returns_amt := COALESCE(
        entity_id.RELATED(
            SUM(tx_amount)
            FILTER (WHERE tx_account = 'returns' AND tx_month = month_input)
            GROUP BY (tx_entity_id)
        ),
        0e0
    ),
    revenue := gross + IF(includes_shipping, shipping, 0e0)
    + IF(includes_returns, returns_amt, 0e0)
RETURN
    entity_name,
    revenue
FROM FM.FINANCE
FOR
    month_input := BIND_VALUE('2024-01'),
    entity_id := BIND_VALUES(ARRAY(2, 3)),
    NESTED tx_id := BIND_COLUMNS(
        tx_id
        FROM SQL(SELECT tx_id FROM tutorial_finance.transactions ORDER BY tx_id)
    )
ORDER BY entity_name
;
```

France Unit reports **29000 EUR**: `30000 - 1000` of returns, with shipping ignored. Germany Unit reports **42500 EUR**: `40000 + 2500` of shipping, with returns ignored.

Both rows come from the same feature definition. The entity's mapped policy determines which components count.

## Reuse the definition in February

Change only the month binding:


```python
%%featureql --client client

DEFINE
    gross := COALESCE(
        entity_id.RELATED(
            SUM(tx_amount)
            FILTER (WHERE tx_account = 'gross_sales' AND tx_month = month_input)
            GROUP BY (tx_entity_id)
        ),
        0e0
    ),
    shipping := COALESCE(
        entity_id.RELATED(
            SUM(tx_amount)
            FILTER (WHERE tx_account = 'shipping_revenue'
            AND tx_month = month_input)
            GROUP BY (tx_entity_id)
        ),
        0e0
    ),
    returns_amt := COALESCE(
        entity_id.RELATED(
            SUM(tx_amount)
            FILTER (WHERE tx_account = 'returns' AND tx_month = month_input)
            GROUP BY (tx_entity_id)
        ),
        0e0
    ),
    revenue := gross + IF(includes_shipping, shipping, 0e0)
    + IF(includes_returns, returns_amt, 0e0)
RETURN
    entity_name,
    revenue
FROM FM.FINANCE
FOR
    month_input := BIND_VALUE('2024-02'),
    entity_id := BIND_VALUES(ARRAY(2, 3)),
    NESTED tx_id := BIND_COLUMNS(
        tx_id
        FROM SQL(SELECT tx_id FROM tutorial_finance.transactions ORDER BY tx_id)
    )
ORDER BY entity_name
;
```

France Unit now reports **27200 EUR** and Germany Unit **48000 EUR**. The local formula and policy mappings are unchanged.

## Roll January up to Europe

At the parent level, first sum the leaf results. January has no intercompany rows, so the gross and consolidated totals are the same:


```python
%%featureql --client client

DEFINE
    gross := COALESCE(
        entity_id.RELATED(
            SUM(tx_amount)
            FILTER (WHERE tx_account = 'gross_sales' AND tx_month = month_input)
            GROUP BY (tx_entity_id)
        ),
        0e0
    ),
    shipping := COALESCE(
        entity_id.RELATED(
            SUM(tx_amount)
            FILTER (WHERE tx_account = 'shipping_revenue'
            AND tx_month = month_input)
            GROUP BY (tx_entity_id)
        ),
        0e0
    ),
    returns_amt := COALESCE(
        entity_id.RELATED(
            SUM(tx_amount)
            FILTER (WHERE tx_account = 'returns' AND tx_month = month_input)
            GROUP BY (tx_entity_id)
        ),
        0e0
    ),
    revenue := gross + IF(includes_shipping, shipping, 0e0)
    + IF(includes_returns, returns_amt, 0e0),
    consolidated_revenue_eur := SUM(revenue) PER entity_id
RETURN
    consolidated_revenue_eur
FROM FM.FINANCE
FOR
    month_input := BIND_VALUE('2024-01'),
    entity_id := BIND_VALUES(ARRAY(2, 3)),
    NESTED tx_id := BIND_COLUMNS(
        tx_id
        FROM SQL(SELECT tx_id FROM tutorial_finance.transactions ORDER BY tx_id)
    )
;
```

The Europe result is **71500 EUR**, exactly `29000 + 42500`.

## Eliminate February's internal sales

February's leaf total is **75200 EUR**, but it includes two sales between France Unit and Germany Unit: `8000` and `6000`. Those transactions are real for the leaves and invalid as external Europe revenue.


```python
%%featureql --client client

DEFINE
    gross := COALESCE(
        entity_id.RELATED(
            SUM(tx_amount)
            FILTER (WHERE tx_account = 'gross_sales' AND tx_month = month_input)
            GROUP BY (tx_entity_id)
        ),
        0e0
    ),
    shipping := COALESCE(
        entity_id.RELATED(
            SUM(tx_amount)
            FILTER (WHERE tx_account = 'shipping_revenue'
            AND tx_month = month_input)
            GROUP BY (tx_entity_id)
        ),
        0e0
    ),
    returns_amt := COALESCE(
        entity_id.RELATED(
            SUM(tx_amount)
            FILTER (WHERE tx_account = 'returns' AND tx_month = month_input)
            GROUP BY (tx_entity_id)
        ),
        0e0
    ),
    revenue := gross + IF(includes_shipping, shipping, 0e0)
    + IF(includes_returns, returns_amt, 0e0),
    interco_elim := COALESCE(
        entity_id.RELATED(
            SUM(tables.interco[amount])
            FILTER (WHERE tables.interco[account] = 'interco_sale'
            AND ic_month = month_input)
            GROUP BY (tables.interco[to_entity_id])
        ),
        0e0
    ),
    consolidated_revenue_eur := SUM(revenue - interco_elim) PER ROW
RETURN
    consolidated_revenue_eur
FROM FM.FINANCE
FOR CROSS
    month_input := BIND_VALUE('2024-02'),
    entity_id := BIND_VALUES(ARRAY(2, 3)),
    NESTED tx_id := BIND_COLUMNS(
        tx_id
        FROM SQL(SELECT tx_id FROM tutorial_finance.transactions ORDER BY tx_id)
    ),
    NESTED interco_id := BIND_COLUMNS(
        interco_id
        FROM SQL(SELECT interco_id FROM tutorial_finance.intercompany ORDER BY interco_id)
    )
;
```

The consolidated result is:

`75200 - (8000 + 6000) = 61200 EUR`

This is why eliminations are non-additive: summing correct leaf measures is not sufficient at the parent grain. See [Metric additivity](https://featuremesh.com/docs/featureql/semantic_modeling/metric_summarizability) for the distinction between a legal rollup and a new target-grain formula.
